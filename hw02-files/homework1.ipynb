```python
min_val = float("inf")
max_val = float("-inf")
total = 0
count = 0

with open("numbers.txt", "r", encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        try:
            num = int(line)
            if num < min_val:
                min_val = num
            if num > max_val:
                max_val = num
            total += num
            count += 1
        except:
            print("Пропущена некоректная строка")
if count > 0:
    avg_val = total / count
    with open('result.txt', 'w', encoding='utf-8') as res:
        res.write(f"Минимальное значение: {min_val}\n")
        res.write(f"Максимальное значение: {max_val}\n")
        res.write(f"Среднее арифметическое: {avg_val:.2f}\n")
    print(f"Готово! Обработано {count} чисел. Результаты в result.txt.")
else:
    print("В файле не найдено ни одного корректного числа.")
```

Третья задача

```python
import csv

with open('students.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    
    total_score = 0
    count = 0
    max_score = -1
    best_student = ""
    group_stats = {} 
    
    for row in reader:
        score = int(row['score'])
        name = row['name']
        group = row['group']
        total_score += score
        count += 1
        
        if score > max_score:
            max_score = score
            best_student = name
        if group not in group_stats:
            group_stats[group] = [0, 0]
        group_stats[group][0] += score
        group_stats[group][1] += 1

avg_score = total_score / count if count > 0 else 0

print(f"Средний балл всех студентов: {avg_score:.2f}")
print(f"Студент с максимальным баллом: {best_student} (балл: {max_score})")
print("Средний балл по группам:")
for grp, (sum_s, cnt) in group_stats.items():
    print(f"  Группа {grp}: {sum_s / cnt:.2f}")
```

```python
class BankError(Exception):
    pass

class InvalidAmountError(BankError):
    pass

class InsufficientFundsError(BankError):
    pass

class AccountBlockedError(BankError):
    pass

def withdraw(balance, amount):
    if amount <= 0:
        raise InvalidAmountError("Сумма снятия должна быть положительной")
    
    if balance < 0:
        raise AccountBlockedError("Счет заблокирован, операции невозможны")
        
    if amount > balance:
        raise InsufficientFundsError(f"Недостаточно средств - текущий баланс: {balance}")
        
    return balance - amount

current_balance = 1000

test_amounts = [500, -100, 2000, 300]

for amt in test_amounts:
    print(f"\nПопытка снятия{amt}...")
    try:
        current_balance = withdraw(current_balance, amt)
        print(f"Успешно! Новый баланс: {current_balance}")
    except BankError as e:
        print(f"Банковская ошибка: {e}")
```

```python
import time


class Timer:
    def __enter__(self):
        self.start = time.perf_counter()
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        elapsed = time.perf_counter() - self.start
        print(f"Время выполнения: {elapsed:.6f} сек")
        return False


def some_function():
    return sum(range(1_000_000))


with Timer():
    some_function()
```

```python
import time
from contextlib import contextmanager


@contextmanager
def timer():
    start = time.perf_counter()
    try:
        yield
    finally:
        elapsed = time.perf_counter() - start
        print(f"Время выполнения: {elapsed:.6f} сек")


def some_function():
    return sum(range(1_000_000))


with timer():
    some_function()
```

```python
import re
from collections import Counter

LOG_PATTERN = re.compile(
    r"^(?P<datetime>\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2}) "
    r"(?P<level>INFO|WARNING|ERROR) "
    r"User (?P<user_id>\d+) "
    r"(?P<message>.*)$"
)

levels = Counter()
error_users = set()

with open("app.log", encoding="utf-8") as log_file:
    for line in log_file:
        match = LOG_PATTERN.match(line.rstrip("\n"))
        if not match:
            continue
        level = match.group("level")
        levels[level] += 1
        if level == "ERROR":
            error_users.add(match.group("user_id"))

for level in ("INFO", "WARNING", "ERROR"):
    print(f"{level}: {levels[level]}")
print("Пользователи с ошибками:", sorted(error_users, key=int))
```

```python
import json
import logging
import re
from pathlib import Path

logging.basicConfig(level=logging.INFO, format="%(levelname)s: %(message)s")
logger = logging.getLogger(__name__)

ACCESS_PATTERN = re.compile(
    r"^\S+ \S+ (?P<ip>\d{1,3}(?:\.\d{1,3}){3}) "
    r"(?P<method>[A-Z]+) (?P<url>\S+) (?P<status>\d{3})$"
)


def analyze_access_log(path):
    methods = {}
    status_4xx = 0
    urls = {}
    total = 0
    try:
        with open(path, encoding="utf-8") as f:
            logger.info("Начата обработка файла %s", path)
            for line in f:
                match = ACCESS_PATTERN.match(line.strip())
                if not match:
                    logger.warning("Строка не распознана: %r", line.strip())
                    continue
                total += 1
                method = match.group("method")
                url = match.group("url")
                status = int(match.group("status"))
                methods[method] = methods.get(method, 0) + 1
                urls[url] = urls.get(url, 0) + 1
                if 400 <= status < 500:
                    status_4xx += 1
    except FileNotFoundError:
        logger.error("Файл не найден: %s", path)
        return None
    except PermissionError:
        logger.error("Нет доступа к файлу: %s", path)
        return None
    except OSError as exc:
        logger.error("Ошибка чтения файла %s: %s", path, exc)
        return None

    most_common_url = max(urls, key=urls.get) if urls else None
    stats = {
        "total_requests": total,
        "requests_by_method": methods,
        "client_errors_4xx": status_4xx,
        "most_common_url": most_common_url,
    }
    logger.info("Обработка завершена, обработано записей: %d", total)
    return stats


stats = analyze_access_log("access.log")
if stats is not None:
    Path("access_stats.json").write_text(
        json.dumps(stats, ensure_ascii=False, indent=2), encoding="utf-8"
    )
    logger.info("Статистика сохранена в access_stats.json")
```
