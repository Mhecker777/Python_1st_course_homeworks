Задание 2

```python
max_val = None
min_val = None
count = 0
curr_pos = 0
max_pos = 0


while True:
    val = int(input())
    if val == 0:
        break
    if max_val is None or val > max_val:
        max_val = val
    if min_val is None or val < min_val:
        min_val = val       
    if val % 2 == 0:
        count += 1      
    if val > 0:
        curr_pos += 1
        if curr_pos > max_pos:
            max_pos = curr_pos
    else:
        curr_pos = 0


print(f"Максимальное число {max_val}")
print(f"Минимальное число {min_val}")
print(f"Количество четных чисел {count}")
print(f"Длина самой длинной последовательности положительных чисел {max_pos}")
    
```

Задача 3

```python
def palindrome(s):
    clean = "".join(char.lower() for char in s if char.isalnum())
    left = 0
    right = len(clean) - 1
    check = True
    while left < right:
        if clean[left] != clean[right]:
            check = False
            break
        left += 1
        right -= 1
    return check


tests = []


s = input()
while s != "0":
    tests.append(s)
    s = input()


for text in tests:
    result = palindrome(text)
    print(f"Оригинал {text}")
    print(f"Палиндром {result}\n")
```
